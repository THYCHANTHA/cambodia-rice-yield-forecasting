```python
import pandas as pd
import numpy as np

# Read the main CAS 2024 data and S4_CROP data
cas_data = pd.read_stata('Data/CAS 2024.dta')
s4_crop_data = pd.read_stata('Data/S4_CROP.dta')

print('=== FOCUSING ON RICE CROPS (101, 102, 103) ===')

# Filter for rice crops only
rice_codes = [101, 102, 103]
rice_data = s4_crop_data[s4_crop_data['CROPCODE'].isin(rice_codes)].copy()

print('Found', len(rice_data), 'rice crop records')
print('Rice crop codes found:', sorted(rice_data['CROPCODE'].unique()))

# Handle categorical data by converting to numeric
numeric_cols = [
    'AREA_PLANTEDHA_TOTAL',
    'AREA_HARVESTEDHA_TOTAL',
    'PRODUCTION_TOTAL',
    'YIELD_TOTAL'
]

for col in numeric_cols:
    if col in rice_data.columns:
        rice_data[col] = pd.to_numeric(rice_data[col], errors='coerce')

# Create comprehensive rice-focused dataset
rice_summary = (
    rice_data.groupby(['interview__key', 'CROPCODE'], observed=True)
    .agg({
        'AREA_PLANTEDHA_TOTAL': lambda x: x.sum(skipna=True),
        'AREA_HARVESTEDHA_TOTAL': lambda x: x.sum(skipna=True),
        'PRODUCTION_TOTAL': lambda x: x.sum(skipna=True),
        'YIELD_TOTAL': lambda x: x.mean(skipna=True)
    })
    .reset_index()
)

# Map crop codes to names
crop_names = {
    101: 'Non-aromatic paddy',
    102: 'Aromatic paddy',
    103: 'Sticky paddy'
}
rice_summary['CROP_NAME'] = rice_summary['CROPCODE'].map(crop_names)

# Merge with household data
rice_household = pd.merge(
    rice_summary,
    cas_data[[
        'interview__key', 'interview__id', 'VID', 'Holding_ID',
        'PROVINCE', 'DISTRICT', 'COMMUNE', 'VILLAGE'
    ]],
    on='interview__key',
    how='left'
)

# Save rice-focused data
rice_household.to_excel('Rice_Crops_Focus.xlsx', index=False)

# Create detailed rice production analysis
rice_detailed = rice_data[
    ['interview__key', 'interview__id', 'CROPCODE'] + numeric_cols
].copy()
rice_detailed['CROP_NAME'] = rice_detailed['CROPCODE'].map(crop_names)
rice_detailed.to_excel('Rice_Detailed_Production.xlsx', index=False)

print('=== RICE CROP ANALYSIS ===')
print('Total rice records:', len(rice_data))
print('Non-aromatic paddy (101):', len(rice_data[rice_data['CROPCODE'] == 101]), 'records')
print('Aromatic paddy (102):', len(rice_data[rice_data['CROPCODE'] == 102]), 'records')
print('Sticky paddy (103):', len(rice_data[rice_data['CROPCODE'] == 103]), 'records')

print('=== FILES CREATED ===')
print('1. Rice_Crops_Focus.xlsx - Summary by household and rice type')
print('2. Rice_Detailed_Production.xlsx - Detailed rice production data')
print('3. Files include join keys: interview__key, interview__id, VID, Holding_ID')

```

```python
import pandas as pd
from pathlib import Path

# === Paths ===
s4_path = Path("data/S4_CROP_cleaned.csv")
dta_path = Path("data/CAS 2024.dta")
output_path = Path("data/CAS2024_merged.csv")

# === Load Section 4 (already cleaned) ===
s4 = pd.read_csv(s4_path)

# === Load full dataset (.dta) for other sections ===
df = pd.read_stata(dta_path)

# --- Section 1A ---
cols_s1a = ["Holding_ID"] + \
           [c for c in df.columns if c.startswith("S1A_Q01__")] + \
           [c for c in ["S1A_Q06","S1A_Q07","LABOR","adult_members",
                        "S1A_Q08","S1A_Q09","S1A_Q10","S1A_Q11","S1A_Q12"] if c in df.columns]
s1a = df[cols_s1a]

# --- Section 2 ---
cols_s2 = ["Holding_ID"] + \
          [c for c in df.columns if c.startswith("S2_Q01__")] + \
          [c for c in ["S2_Q02","S2_Q03","S2_Q05"] if c in df.columns]
s2 = df[cols_s2]

# --- Section 3 ---
cols_s3 = ["Holding_ID"]
cols_s3 += [c for c in ["S3_Q03","S3_Q04","S3_Q04_unit","S3_Q05","S3_Q06",
                        "TOTALPARCELUTILIZED","AAU","TOTALAREA"] if c in df.columns]
s3 = df[cols_s3]

# --- Section 11 ---
cols_s11 = ["Holding_ID"] + \
           [c for c in df.columns if c.startswith("S11_Q05__")] + \
           [c for c in df.columns if c.startswith("S11_Q14")] + \
           [c for c in df.columns if c.startswith("S11_Q15")] + \
           [c for c in df.columns if c.startswith("S11_Q20")] + \
           [c for c in df.columns if c.startswith("S11_Q21")] + \
           [c for c in df.columns if c.startswith("S11_Q22")] + \
           [c for c in df.columns if c.startswith("S11_Q23")] + \
           [c for c in df.columns if c.startswith("S11_Q24")] + \
           [c for c in df.columns if c.startswith("S11_Q25")] + \
           [c for c in df.columns if c.startswith("S11_Q26")] + \
           [c for c in df.columns if c.startswith("S11_Q27")]
s11 = df[cols_s11]

# --- Section 12 ---
cols_s12 = ["Holding_ID"] + \
           [c for c in df.columns if c.startswith("S12_Q01__")] + \
           [c for c in df.columns if c.startswith("S12_Q03__")] + \
           [c for c in df.columns if c.startswith("S12_Q15__")] + \
           [c for c in df.columns if c.startswith("S12_Q21__")]
s12 = df[cols_s12]

# --- Section 13 ---
cols_s13 = ["Holding_ID","mostsevere"] + \
           [c for c in df.columns if c.startswith("S13_Q02__")] + \
           [c for c in ["S13_Q08","S13_Q09","S13_Q10","S13_Q11","S13_Q12"] if c in df.columns] + \
           [c for c in df.columns if c.startswith("S13_Q14__")]
s13 = df[cols_s13]

# === Merge all sections into one (left joins) ===
merged = (
    s4.merge(s1a, left_on="hold_id", right_on="Holding_ID", how="left")
      .merge(s2, on="Holding_ID", how="left")
      .merge(s3, on="Holding_ID", how="left")
      .merge(s11, on="Holding_ID", how="left")
      .merge(s12, on="Holding_ID", how="left")
      .merge(s13, on="Holding_ID", how="left")
)

# === Drop duplicate Holding_ID (since s4 already has hold_id) ===
merged = merged.drop(columns=["Holding_ID"])

# === Save final merged dataset ===
merged.to_csv(output_path, index=False, encoding="utf-8")
print(f"✅ Final merged dataset saved: {output_path}")
```
