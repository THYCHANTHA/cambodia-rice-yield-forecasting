```python
import pandas as pd

# File paths
main_file = "data/final_visualization.csv"
crop_file = "data/S4_CROP_cleaned.csv"
output_file = "data/merged_visualization.csv"

# Load datasets
main_df = pd.read_csv(main_file)
crop_df = pd.read_csv(crop_file)

# Select only required columns from crop file
crop_selected = crop_df[["interview__key", "prov", "dist", "comm", "vill"]]

# Merge on 'interview__key'
merged_df = pd.merge(main_df, crop_selected, on="interview__key", how="left")

# Save new file
merged_df.to_csv(output_file, index=False, encoding="utf-8")

print("✅ Merged file saved at:", output_file)
```

```python
import pandas as pd
import plotly.express as px

# Assuming df has 'prov' and 'zone'
# Aggregate counts
zone_prov_counts = df.groupby('zone')['prov'].nunique().reset_index()
zone_prov_counts.columns = ['zone', 'province_count']

# Provinces list per zone
provinces_per_zone = df.groupby('zone')['prov'].apply(lambda x: ', '.join(sorted(x.unique()))).reset_index()
zone_prov_counts = zone_prov_counts.merge(provinces_per_zone, on="zone")
zone_prov_counts.rename(columns={'prov': 'Provinces'}, inplace=True)

# Interactive bar chart
fig = px.bar(
    zone_prov_counts,
    x="zone",
    y="province_count",
    color="zone",
    text="province_count",
    hover_data={"Provinces": True, "zone": False, "province_count": False}  # show only province names on hover
)

fig.update_layout(
    title="Provinces per Agro-Ecological Zone in Cambodia",
    xaxis_title="Agro-Ecological Zone",
    yaxis_title="Number of Provinces",
    showlegend=False
)

fig.show()
```

```python
import pandas as pd

# Function to check and optionally fix mojibake (safer approach)
def fix_mojibake(text):
    if isinstance(text, str):
        try:
            # Test if text is already valid UTF-8 by encoding/decoding
            text.encode("utf-8").decode("utf-8")
            return text  # No fix needed; text is likely fine
        except UnicodeEncodeError:
            try:
                # Try to fix mojibake by assuming latin1-encoded text
                return text.encode("latin1", errors="ignore").decode("utf-8", errors="ignore")
            except UnicodeDecodeError:
                return text  # Return original if both fail
    return text

# Columns with Khmer text
khmer_cols = ["dist", "comm", "vill"]

# Apply mojibake fix and create lowercase columns
for col in khmer_cols:
    if col in df.columns:
        # Apply fix only if necessary
        df[col] = df[col].apply(fix_mojibake)
        # Create lowercase duplicate for search/join
        df[col.lower() + "_km"] = df[col].astype(str).str.strip().str.lower()
    else:
        print(f"Column '{col}' not found in DataFrame!")

# Save to CSV (UTF-8 encoding with BOM for Excel compatibility)
output_file = "final_visualization_crop.csv"
df.to_csv(output_file, index=False, encoding="utf-8-sig")

# Verify the save
print(f"✅ Cleaned data saved to {output_file}")
print(f"Shape of saved dataset: {df.shape}")
print(f"Columns in saved dataset: {df.columns.tolist()}")
print("\nFirst few rows of Khmer columns:")
print(df[["dist", "comm", "vill"] + [col + "_km" for col in khmer_cols]].head())

# Optional: Summary of non-empty Khmer text
print("\nSummary of non-empty Khmer text:")
for col in khmer_cols:
    if col in df.columns:
        non_empty = df[col].dropna().count()
        print(f"{col}: {non_empty} non-empty entries")
```
